```python
# Данные: заказы за неделю 
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
```

## [База] 
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.

```python
cities_clean = []
for c in cities:
    cities_clean.append(c.strip().lower().title())
print(cities_clean)
```

2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.

```python
amounts_int = []
for a in amounts:
    amounts_int.append(int(a))
print(amounts_int)
```

3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).

```python
print(f'Общее число заказов: {len(amounts_int)}')
print(f'Общая сумма: {sum(amounts_int)}')
print(f'Средний чек: {round(sum(amounts_int) / len(amounts_int))}')
```

4. Сколько заказов не оплачено и на какую сумму?

```python
not_paid = []
for p, a in zip(paid, amounts_int):
    if not p:
        not_paid.append(a)
print(f'Количество неоплаченных заказов: {len(not_paid)}')
print(f'Сумма неоплаченных заказов: {sum(not_paid)}')
```

5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

```python
print("Город | Сумма | Статус")
for city, amount, status in zip(cities_clean, amounts_int, paid):
    print(f"{city} | {amount} | {'Оплачен' if status else 'Неоплачен'}")
```

## [Стандарт]

6. Выручка (только оплаченные) по Алматы и по Астане отдельно.

```python
ata = []
ast = []
for c, a, p in zip(cities_clean, amounts_int, paid):
    if c == 'Almaty' and p:
        ata.append(a)
    elif c == 'Astana' and p:
        ast.append(a)
print(f'Выручка по Алмате: {sum(ata)}')
print(f'Выручка по Астане: {sum(ast)}')
```

7. Номер (индекс) и город самого крупного **оплаченного** заказа.

```python
max_amount = 0
max_index = None
max_city = None
for i, (c, a, p) in enumerate(zip(cities_clean, amounts_int, paid)):
    if p and a > max_amount:
        max_amount = a
        max_index = i 
        max_city = c
print(f'Индекс: {max_index}')
print(f'Город: {max_city}')
print(f'Самый крупный оплаченный заказ: {max_amount}')
```

8. Доля заказов выходного дня (`сб`, `вс`) в процентах.

```python
demalys = []
for a, d in zip(amounts_int, days):
    if d == 'сб' or d == 'вс':
        demalys.append(a)
print(f'Доля заказов выходного дня: {(len(demalys) / len(amounts_int)) * 100}')
print(f'Выходных заказов {len(demalys)} из {len(amounts_int)}')
```

9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

```python
savings = 0
months = 0
while savings < 400000:
    savings += 15000
    months += 1
print(f'Спустя {months} месяцев откладывая ежемесячно по 15000, накопится {savings}тг')
```

```python
savings2 = 0
months2 = 0
monthly_savings = 15000
while savings2 < 400000:
    savings2 += monthly_savings
    monthly_savings += 1000
    months2 += 1
print(f'Спустя {months2} месяцев, ежемесячно откладывая 15000 и прибавляя 1000, накопится: {savings2}')
```

## [Челлендж]

10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.

```python
unique_cities = []
revenue_by_city = []
for city in cities_clean:
    if city not in unique_cities:
        unique_cities.append(city)
        
for city in unique_cities:
    revenue = 0
    for c, a, p in zip(cities_clean, amounts_int, paid):
        if city == c and p:
            revenue += a
    revenue_by_city.append(revenue)
print(sorted(zip(unique_cities, revenue_by_city), key=lambda x: x[1], reverse=True))
```

11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.

```python
sorted_amounts = sorted(amounts_int)
n = len(sorted_amounts)
if n % 2 == 0:
    median = (sorted_amounts[n // 2 - 1] + sorted_amounts[n // 2]) / 2
else:
    median = sorted_amounts[n // 2]
print(f'Медиана: {median}')
```

12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.

```python
unique_days = []
for day in days:
    if day not in unique_days:
        unique_days.append(day)
for day in unique_days:
    revenue = 0
    for d, a, p in zip(days, amounts_int, paid):
        if day == d and p:
            revenue += a
    print(f'{day}: {"█" * (revenue // 5000)} ({revenue} тг)')
```
